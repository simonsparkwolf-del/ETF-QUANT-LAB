```python
import pandas as pd
from QuantLab.utils.config import load_pathes
from pathlib import Path
from datetime import date
```

```python
folder = load_pathes()["data"]
```

```python
panel_path = folder/"raw"/"2000_2026"/"ETF V01.xlsx"
spy_path = folder/"raw"/"2000_2026"/"SPY_2020.xlsx"
```

```python
panel = pd.read_excel(panel_path)
spy = pd.read_excel(spy_path)
```

```python
panel.head()
```

```python
panel.iloc[2,:] = panel.iloc[2,:].ffill()
```

```python
name = panel.iloc[2,:].fillna('')
cat = panel.iloc[4,:]
columns = [(i+' '+j).strip() for i,j in zip(name,cat)]
```

```python
columns
```

```python
panel = panel.iloc[5:,:]
panel.columns = columns
```

```python
# panel.index = panel["Dates"]
```

```python
spy.head()
```

```python
spy.iloc[2,:] = spy.iloc[2,:].ffill()
```

```python
name = spy.iloc[2,:].fillna('')
cat = spy.iloc[4,:]
columns = [(i+' '+j).strip() for i,j in zip(name,cat)]
```

```python
spy = spy.iloc[5:,:]
spy.columns = columns
```

```python
# spy.index = spy["Dates"]
```

```python
spy.head()
```

```python
processed = pd.merge(spy,panel,how="inner",on="Dates")
```

```python
processed.head()
```

```python
output_path = folder/"processed"/"data.csv"
```

```python
processed.to_csv(output_path,index=False)
```
